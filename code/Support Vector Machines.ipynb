```python
credit_card_data = credit_card_data.drop("Time", axis=1)
```

```python
from sklearn import preprocessing
scaler = preprocessing.StandardScaler()
```

```python
#standard scaling
credit_card_data['std_Amount'] = scaler.fit_transform(credit_card_data['Amount'].values.reshape (-1,1))

#removing Amount
credit_card_data = credit_card_data.drop("Amount", axis=1)
```

```python
sns.countplot(x="Class", data=credit_card_data)
```

```python
import imblearn 
from imblearn.under_sampling import RandomUnderSampler 

undersample = RandomUnderSampler(sampling_strategy=0.5)
```

```python
cols = credit_card_data.columns.tolist()
cols = [c for c in cols if c not in ["Class"]]
target = "Class"
```

```python
# Split BEFORE undersampling, then undersample only the training fold.
# (Previously the whole dataset was undersampled first and THEN split,
# which lets the resampling step see rows that end up in the test set.)
X = credit_card_data[cols]
Y = credit_card_data[target]
```

```python
from sklearn.model_selection import train_test_split

# Stratified split on the ORIGINAL (still-imbalanced) data first.
X_train_full, X_test, y_train_full, y_test = train_test_split(
    X, Y, test_size=0.2, stratify=Y, random_state=1
)

# Undersample only the training fold -- the test set stays untouched and
# keeps the real-world class balance, which is what a model faces in production.
X_train, y_train = undersample.fit_resample(X_train_full, y_train_full)
```

```python
from pandas import DataFrame
test = pd.DataFrame(y_train, columns=['Class'])
```

```python
#visualizing undersampling results (training fold only -- test set is untouched)
fig, axs = plt.subplots(ncols=2, figsize=(13,4.5))
sns.countplot(x=y_train_full, ax=axs[0])
sns.countplot(x="Class", data=test, ax=axs[1])

fig.suptitle("Class repartition before and after undersampling (train fold)")
a1=fig.axes[0]
a1.set_title("Before")
a2=fig.axes[1]
a2.set_title("After")
```
